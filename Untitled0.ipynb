```python
from flask import Flask, request, redirect, session, render_template_string
from google.colab import output
import sqlite3
import datetime

app = Flask(__name__)
app.secret_key = "event_secret_key"

# ---------------- DATABASE ---------------- #

def get_db():
    return sqlite3.connect("event.db")

def init_db():
    db = get_db()
    c = db.cursor()

    c.execute("""CREATE TABLE IF NOT EXISTS membership(
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        vendor TEXT,
        plan TEXT,
        expiry TEXT
    )""")

    c.execute("""CREATE TABLE IF NOT EXISTS products(
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT,
        price INTEGER
    )""")

    c.execute("""CREATE TABLE IF NOT EXISTS cart(
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        product TEXT,
        price INTEGER,
        qty INTEGER
    )""")

    c.execute("""CREATE TABLE IF NOT EXISTS orders(
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT,
        total INTEGER,
        status TEXT
    )""")

    db.commit()
    db.close()

init_db()

# ---------------- LOGIN ---------------- #

login_html = """
<h2>Event Management System</h2>

<form method="POST">
User ID:
<input type="text" name="username" required><br><br>

Password:
<input type="password" name="password" required><br><br>

Role:
<input type="radio" name="role" value="admin" required> Admin
<input type="radio" name="role" value="vendor"> Vendor
<input type="radio" name="role" value="user"> User
<br><br>

<button type="submit">Login</button>
</form>
"""

@app.route("/", methods=["GET","POST"])
def login():
    if request.method == "POST":
        role = request.form.get("role")

        # SAFETY CHECK (PREVENT SILENT FAIL)
        if not role:
            return "<h3>Please select a role</h3><a href='/'>Back</a>"

        session["username"] = request.form.get("username")
        session["role"] = role

        return redirect("/dashboard")

    return render_template_string(login_html)

# ---------------- DASHBOARD ---------------- #

@app.route("/dashboard")
def dashboard():
    role = session.get("role")

    if role == "admin":
        return """
        <h2>Welcome Admin</h2>
        <a href='/membership'>Membership</a><br><br>
        <a href='/logout'>Logout</a>
        """

    if role == "vendor":
        return """
        <h2>Welcome Vendor</h2>
        <a href='/add_product'>Add Product</a><br><br>
        <a href='/logout'>Logout</a>
        """

    if role == "user":
        return """
        <h2>Welcome User</h2>
        <a href='/products'>View Products</a><br>
        <a href='/cart'>Cart</a><br>
        <a href='/order_status'>Order Status</a><br><br>
        <a href='/logout'>Logout</a>
        """

    return redirect("/")

# ---------------- MEMBERSHIP ---------------- #

@app.route("/membership", methods=["GET","POST"])
def membership():
    if session.get("role") != "admin":
        return "Access Denied"

    if request.method == "POST":
        vendor = request.form["vendor"]
        plan = request.form["plan"]
        expiry = (datetime.date.today() + datetime.timedelta(days=180)).isoformat()

        db = get_db()
        db.execute(
            "INSERT INTO membership(vendor,plan,expiry) VALUES(?,?,?)",
            (vendor,plan,expiry)
        )
        db.commit()
        db.close()

        return "Membership Added<br><a href='/dashboard'>Back</a>"

    return """
    <h3>Add Membership</h3>
    <form method="post">
    Vendor Name: <input required name="vendor"><br><br>

    Plan:
    <input type="radio" name="plan" value="6 Months" checked> 6 Months
    <input type="radio" name="plan" value="1 Year"> 1 Year
    <input type="radio" name="plan" value="2 Years"> 2 Years
    <br><br>

    <input type="submit">
    </form>
    """

# ---------------- PRODUCTS ---------------- #

@app.route("/add_product", methods=["GET","POST"])
def add_product():
    if session.get("role") != "vendor":
        return "Access Denied"

    if request.method == "POST":
        name = request.form["name"]
        price = int(request.form["price"])

        db = get_db()
        db.execute(
            "INSERT INTO products(name,price) VALUES(?,?)",
            (name,price)
        )
        db.commit()
        db.close()

        return redirect("/dashboard")

    return """
    <h3>Add Product</h3>
    <form method="post">
    Name: <input required name="name"><br><br>
    Price: <input required name="price"><br><br>
    <input type="submit">
    </form>
    """

@app.route("/products")
def products():
    db = get_db()
    rows = db.execute("SELECT * FROM products").fetchall()
    db.close()

    html = "<h3>Products</h3>"
    for r in rows:
        html += f"{r[1]} - ₹{r[2]} <a href='/add_to_cart/{r[1]}/{r[2]}'>Add</a><br>"
    html += "<br><a href='/dashboard'>Back</a>"
    return html

# ---------------- CART ---------------- #

@app.route("/add_to_cart/<name>/<int:price>")
def add_to_cart(name, price):
    db = get_db()
    db.execute(
        "INSERT INTO cart(product,price,qty) VALUES(?,?,1)",
        (name,price)
    )
    db.commit()
    db.close()
    return redirect("/cart")

@app.route("/cart")
def cart():
    db = get_db()
    rows = db.execute("SELECT * FROM cart").fetchall()
    db.close()

    total = sum(r[2]*r[3] for r in rows)

    html = "<h3>Cart</h3>"
    for r in rows:
        html += f"{r[1]} x {r[3]} = ₹{r[2]*r[3]}<br>"
    html += f"<br><b>Total = ₹{total}</b>"
    html += "<br><a href='/checkout'>Checkout</a>"
    return html

# ---------------- CHECKOUT ---------------- #

@app.route("/checkout", methods=["GET","POST"])
def checkout():
    if request.method == "POST":
        name = request.form["name"]

        db = get_db()
        rows = db.execute("SELECT * FROM cart").fetchall()
        total = sum(r[2]*r[3] for r in rows)

        db.execute(
            "INSERT INTO orders(name,total,status) VALUES(?,?,?)",
            (name,total,"Received")
        )
        db.execute("DELETE FROM cart")
        db.commit()
        db.close()

        return f"""
        <h3>Order Successful</h3>
        Total Amount: ₹{total}<br><br>
        <a href='/dashboard'>Home</a>
        """

    return """
    <h3>Checkout</h3>
    <form method="post">
    Name: <input required name="name"><br><br>

    Payment Method:
    <select>
        <option>Cash</option>
        <option>UPI</option>
    </select><br><br>

    <input type="submit" value="Order Now">
    </form>
    """

# ---------------- ORDER STATUS ---------------- #

@app.route("/order_status")
def order_status():
    db = get_db()
    rows = db.execute("SELECT * FROM orders").fetchall()
    db.close()

    html = "<h3>Order Status</h3>"
    for r in rows:
        html += f"{r[1]} - ₹{r[2]} - {r[3]}<br>"
    html += "<br><a href='/dashboard'>Back</a>"
    return html

# ---------------- LOGOUT ---------------- #

@app.route("/logout")
def logout():
    session.clear()
    return redirect("/")

# ---------------- RUN (COLAB SAFE) ---------------- #

PORT = 5002
output.serve_kernel_port_as_iframe(PORT)
app.run(host="0.0.0.0", port=PORT)
```
